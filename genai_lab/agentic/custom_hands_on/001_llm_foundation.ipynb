An LLM application typically looks like:

``` User → LLM → Response ```

An agent adds the **ability to decide and act**:

An agent is an LLM-driven system that can choose actions, execute them, observe results, and continue until it can answer or complete the task.

# Agentic Development — Step 1

## Goal

Understand the fundamental building block of an agent:

User → LLM → Response

Before building agents, we first need to understand how to call an LLM programmatically.

````python
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
````

````python
# os.environ["OPENAI_API_KEY"] = "YOUR_API_KEY" # getting loaded from .env
client = OpenAI()
response = client.responses.create(
    model="gpt-4.1-mini",
    input="What is an AI agent? Explain in one sentence."
)

print(response.output_text)
````

## Our first LLM abstraction

Instead of directly calling the API every time, create a function.

This will become the foundation on which we later build an agent.

````python
def ask_llm(prompt):
    response = client.responses.create(
        model="gpt-4.1-mini",
        input=prompt
    )
    
    return response.output_text
````

````python
answer = ask_llm("What is tool calling?")
print(answer)
````

````python
ask_llm("What is 25 * 17?")
````

````python
ask_llm("What is the current weather in Delhi?")
````
