# Forward Pass | Making A Guess

## Simple Linear Regression

Linear regression predicts a value using:

$$
\hat{y} = XW + b
$$

- **X** → input features
- **W** → weights learned by the model
- **b** → bias/intercept
- **ŷ** → predicted output

In simple terms:

> **Prediction = input × weight + bias**

W and b are two knob our model adjust

 ![The 5-step deep learning recipe](torch_5_step.png)

````python
import torch

# Number of data points in our dataset
N = 10

# Each data point has:
# D_in  = number of input features
# D_out = number of output values
D_in = 1
D_out = 1

# Create random input data X
# Shape: [N, D_in] = [10, 1]
X = torch.randn(N, D_in)
print(X)
````

````python
# Create the "true" parameters used to generate our target values
# We want to create fake data that follows: y = 2x + 1
# The true relationship is: y = 2x + 1
true_W = torch.tensor([[2.0]])   # True weight
true_b = torch.tensor(1.0)       # True bias

# Generate the target values y
# X @ true_W → linear transformation (2x)
# + true_b   → adds the bias (+1)
# + random noise → makes the data slightly imperfect
y_true = X @ true_W + true_b + torch.randn(N, D_out) * 0.1

# The generated data approximately follows:
# y = 2x + 1
# with a small amount of random noise

print(y_true)
````

# the parameters : models brain

````python
# Initialize our parameters with random values.
# Shapes must be correct for matrix multiplication!

W = torch.randn(D_in, D_out, requires_grad=True)
b = torch.randn(1, requires_grad=True)

# the model intial hypothesis which is wrong but its ok, its start
print(f"Initial Weight W:\n {W}\n")
print(f"Initial Bias b:\n {b}")
````

## Implementation: From Math to Code

### The Math

$$
\hat{y} = XW + b
$$

### The Code

```python
y_hat = X @ W + b

```
# Thats it, Thats the forward pass

````python
import torch

# --------------------------------------------------
# Model implementation
# --------------------------------------------------

def model(X):
    # Forward pass:
    # prediction = XW + b
    return X @ W + b


# --------------------------------------------------
# Make predictions using our model
# --------------------------------------------------

y_hat = model(X)

# Compare predictions with the actual target values
print(f"Prediction y_hat (first 3 rows):\n{y_hat[:3]}\n")

print(f"True Labels y_true (first 3 rows):\n{y_true[:3]}")
````
